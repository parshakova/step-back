# CIFAR-10 / ResNet-20 
# Train only matrix params 
## Compare spectral-clip and spectral-inv variants

```python
from stepback.plot_utils import load_sweep, plot_val_acc_vs_lr, plot_val_acc_vs_epoch
import pandas as pd
```

```python
R1 = load_sweep("cifar10_resnet20_fix_nonmatrix")
R2 = load_sweep("cifar10_resnet20_fix_nonmatrix_remaining")
R3 = load_sweep("cifar10_resnet20_fix_nonmatrix_muon_bot_p_sweep")
R4 = load_sweep("cifar10_resnet20_fix_nonmatrix_high_lr_spectral_inv")
R5 = load_sweep("cifar10_resnet20_fix_nonmatrix_high_lr")

all_records = [R1, R2, R3, R4, R5]


# Merge: concatenate base_df and id_df, keep R1's aesthetics
R1.base_df = pd.concat([r.base_df for r in all_records], ignore_index=True)
R1.id_df = pd.concat([r.id_df for r in all_records])
R1.id_df = R1.id_df[~R1.id_df.index.duplicated(keep="first")]

# Rename muon-bot-p variants by their bottom_p value
def _rename_bot_p(row):
    if row["name"] == "muon-bot-p" and "bottom_p" in row.index and row["bottom_p"] != "none":
        return f"muon-bot-p{int(float(row['bottom_p'])*100)}"
    return row["name"]

R1.id_df["name"] = R1.id_df.apply(_rename_bot_p, axis=1)
R1.base_df = R1.base_df.merge(R1.id_df[["name"]], how="left", left_on="id", right_index=True, suffixes=("_old", ""))
R1.base_df["name"] = R1.base_df["name"].fillna(R1.base_df["name_old"])
R1.base_df = R1.base_df.drop(columns=["name_old"])

R = R1
R.id_df
```

```python
fig, ax = plot_val_acc_vs_lr(R, filename="output/cifar10_resnet20_fix_nonmatrix_val_acc_vs_lr.pdf",
                             skip="sgd-m", ylim=[0, 0.95])
```

```python
fig, ax = plot_val_acc_vs_epoch(R, ylim=[0.8, 0.9])
```

```python
# ── Best config per method: val accuracy & val loss ──
from stepback.plot_utils import _display_name

rows = []
for name in R.id_df["name"].unique():
    sub = R.base_df[R.base_df["name"] == name]
    final = sub[sub["epoch"] == sub["epoch"].max()]
    best_row = final.loc[final["val_score"].idxmax()]
    conf = dict(item.split("=") for item in best_row["id"].split(":"))
    rows.append({
        "Method": _display_name(name),
        "lr": float(conf["lr"]),
        "Val Acc": f'{best_row["val_score"]:.4f}',
        "Val Loss": f'{best_row["val_loss"]:.4f}',
    })

table = pd.DataFrame(rows).sort_values("Val Acc", ascending=False).reset_index(drop=True)
table
```
